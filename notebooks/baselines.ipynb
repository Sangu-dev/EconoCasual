```python
import sys; sys.path.append("..")
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df=pd.read_csv('../data/retail_campaign.csv')
df.head()
```

```python
from sklearn.linear_model import LinearRegression

features = [c for c in df.columns if c not in ('customer_id', 'discount', 'revenue')]
naive = df[df.discount==1].revenue.mean() - df[df.discount==0].revenue.mean()
lr = LinearRegression().fit(df[features + ['discount']], df['revenue'])

pd.DataFrame({'method': ['Naive', 'Linear regression + controls'],
              'estimated_effect': [naive, lr.coef_[-1]]}).round(4)
```
